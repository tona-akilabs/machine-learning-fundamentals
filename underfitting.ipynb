```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
np.random.seed(42)

# House size (m²)
X = 50 + 150 * np.random.rand(50, 1)

# True relationship (nonlinear in reality)
y = 30000 + 2000 * X + 0.05 * (X**2)
y = y.ravel() + np.random.randn(50) * 20000  # noise
```

```python
plt.scatter(X, y)
plt.xlabel("House Size")
plt.ylabel("Price")
plt.title("House Price Dataset")
plt.show()
```

Linear Regression (Underfitting)

```python
model = LinearRegression()
model.fit(X, y)
```

```python
X_test = np.linspace(50, 200, 100).reshape(-1, 1)
y_pred = model.predict(X_test)
```

```python
plt.scatter(X, y)
plt.plot(X_test, y_pred)
plt.title("Linear Regression (Underfitting)")
plt.show()
```

Polynomial Regression (Better)

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline

model = make_pipeline(
    PolynomialFeatures(degree=2),
    LinearRegression()
)

model.fit(X, y)
```

```python
y_pred = model.predict(X_test)
```

```python
plt.scatter(X, y)
plt.plot(X_test, y_pred)
plt.title("Polynomial Degree 2 (Good Fit)")
plt.show()
```

High Degree → Overfitting

```python
model = make_pipeline(
    PolynomialFeatures(degree=10),
    LinearRegression()
)

model.fit(X, y)
```

```python
y_pred = model.predict(X_test)
```

```python
plt.scatter(X, y)
plt.plot(X_test, y_pred)
plt.title("Polynomial Degree 10 (Overfitting)")
plt.show()
```

Detect Overfitting

```python
from sklearn.metrics import mean_squared_error

train_pred = model.predict(X)
train_error = mean_squared_error(y, train_pred)

# Simulate new unseen houses
X_new = np.linspace(50, 200, 100).reshape(-1, 1)
y_true = 30000 + 2000 * X_new + 0.05 * (X_new**2)

test_pred = model.predict(X_new)
test_error = mean_squared_error(y_true, test_pred)

print("Train error:", train_error)
print("Test error:", test_error)
```

Fix with Ridge (Regularization)

```python
from sklearn.linear_model import Ridge

model = make_pipeline(
    PolynomialFeatures(degree=10),
    Ridge(alpha=10)
)

model.fit(X, y)
```

```python
y_pred = model.predict(X_test)
```

```python
plt.scatter(X, y)
plt.plot(X_test, y_pred)
plt.title("Ridge Regression (Fixed Overfitting)")
plt.show()
```
