```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
np.random.seed(0)

X = np.sort(5 * np.random.rand(30, 1), axis=0)
y = np.sin(X).ravel()
y += np.random.randn(30) * 0.2  # noise

plt.scatter(X, y)
plt.title("Dataset")
plt.show()
```

```python
model = LinearRegression()
model.fit(X, y)
```

```python
X_test = np.linspace(0, 5, 100).reshape(-1, 1)
y_pred = model.predict(X_test)

plt.scatter(X, y)
plt.plot(X_test, y_pred)
plt.title("Linear Regression (Underfitting)")
plt.show()
```

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline

model = make_pipeline(
    PolynomialFeatures(degree=2),
    LinearRegression()
)

model.fit(X, y)
```

```python
y_pred = model.predict(X_test)
```

```python
plt.scatter(X, y)
plt.plot(X_test, y_pred)
plt.title("Polynomial Degree 2 (Good Fit)")
plt.show()
```

```python
model = make_pipeline(
    PolynomialFeatures(degree=15),
    LinearRegression()
)

model.fit(X, y)
```

```python
y_pred = model.predict(X_test)
```

```python
plt.scatter(X, y)
plt.plot(X_test, y_pred)
plt.title("Polynomial Degree 15 (Overfitting)")
plt.show()
```

Compare Errors

```python
from sklearn.metrics import mean_squared_error

# Train error
train_pred = model.predict(X)
train_error = mean_squared_error(y, train_pred)
```

```python
# Test data (new unseen)
X_new = np.linspace(0, 5, 100).reshape(-1, 1)
y_true = np.sin(X_new).ravel()
test_pred = model.predict(X_new)
test_error = mean_squared_error(y_true, test_pred)
print("Train error:", train_error)
print("Test error:", test_error)
```

Fix with Ridge Regression

```python
from sklearn.linear_model import Ridge

model = make_pipeline(
    PolynomialFeatures(degree=15),
    Ridge(alpha=1.0)
)
model.fit(X, y)
```

```python
y_pred = model.predict(X_test)
```

```python
plt.scatter(X, y)
plt.plot(X_test, y_pred)
plt.title("Ridge Regression (Fixed Overfitting)")
plt.show()
```
